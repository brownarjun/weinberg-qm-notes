## Eq. 1.2.1–1.2.2 — Rutherford backscatter: when can the target recoil backwards?

```python
import sympy as sp
# 0 = (1+M/m)u^2 - 2(M/m)u - (1 - M/m),  u = v'/v ;  interesting root:
u = -(sp.Symbol('m') - sp.Symbol('M'))/(sp.Symbol('m') + sp.Symbol('M'))
print("v'/v =", u, "  -> negative (backwards recoil) only if m > M  [Eq. 1.2.2]")
```

## Eq. 1.2.7 & 1.2.11 — Bohr energies (symbolic)

```python
n, Z, e, m, hbar = sp.symbols('n Z e m hbar', positive=True)
v = Z*e**2/(n*hbar)                      # from mvr = n*hbar + m v^2/r = Z e^2/r^2
r = n**2*hbar**2/(Z*m*e**2)
E = sp.Rational(1,2)*m*v**2 - Z*e**2/r
print("v =", v)
print("r =", r)
print("E =", sp.simplify(E), "  [Weinberg 1.2.7: -Z^2 e^4 m/(2 n^2 hbar^2)]")
```

## Numerics: hydrogen and He+ spectrum with reduced mass (Eq. 1.2.11)

```python
import numpy as np, sys
sys.path.insert(0, '..')
from weinberg_constants import me, hbar, e, Ryd, h, c

mH, mHe = 1.6735e-24, 6.6464e-24
muH  = me/(1 + me/mH)
muHe = me/(1 + me/mHe)
def E_n(Z, n, mu): return -Ryd*(mu/me)*Z**2/n**2

print(f"H  ground state: {E_n(1,1,muH)/1.602e-12:8.4f} eV")
print(f"He+ ground state: {E_n(2,1,muHe)/1.602e-12:8.4f} eV")
print(f"He+/H ground ratio: {E_n(2,1,muHe)/E_n(1,1,muH):.5f}  [Weinberg: 4.00163]")
# Balmer alpha: n=3 -> 2 in hydrogen
lam = h*c/(E_n(1,3,muH) - E_n(1,2,muH))
print(f"H-alpha wavelength: {lam*1e8:.1f} Angstrom")
```
