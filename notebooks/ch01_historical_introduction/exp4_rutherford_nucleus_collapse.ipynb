# Classical Experiment 4 — Discovery of the Nucleus & the Classical Collapse of the Atom
*(Rutherford scattering: Geiger–Marsden 1909, Rutherford 1911; collapse estimate: background for Weinberg §1.2)*
**What the experiments showed (and where classical physics broke)**

- **Geiger & Marsden (1909),** $\alpha$ particles on gold foil: most pass through, but **~1 in 8000 bounces
  backwards** ($\theta > 90^\circ$). Thomson's diffuse "plum-pudding" atom cannot kick a heavy $\alpha$ backwards —
  it predicts large-angle scattering orders of magnitude too rare (Weinberg's Eq. 1.2.1–1.2.2 argument:
  backward recoil of the target requires the projectile to be *heavier* than the distributed charge it hits).
- **Rutherford (1911):** the positive charge sits in a **tiny, heavy nucleus**; the point-charge Coulomb field
  $E \sim Ze/r^2$ near $r \sim 10^{-12}$ cm flings back a light $\alpha$. The angular distribution follows the
  famous $\csc^4(\theta/2)$ law — confirmed by Geiger & Marsden (1913) over four decades of angle.
- **Contradiction #2 (the one that matters for quantum mechanics):** a Rutherford atom is a miniature solar
  system, and Maxwell's electrodynamics says an orbiting electron must **radiate** (Larmor: $P = 2e^2a^2/3c^3$).
  Energy loss → spiral-in. A hydrogen atom starting at $a_0$ collapses in
  $$\tau = \frac{m_{\rm e}^2 c^3 a_0^3}{4e^4} \approx 1.6\times10^{-11}\ {\rm s},$$
  chirping through a **continuous** spectrum — yet atoms are stable for $10^{10}$+ s and emit **discrete sharp lines**.


```python
%matplotlib inline
import numpy as np, matplotlib.pyplot as plt, sys
sys.path.insert(0, '..')
from weinberg_constants import h, hbar, e, me, c, a0

# --- collapse time from Larmor power + circular-orbit energy E(r) = -e^2/(2r) ---
# |dE/dt| = 2 e^2 a^2/(3 c^3),  a = e^2/(m r^2)  ->  |dr/dt| = 4 e^4/(3 m^2 c^3 r^2)
# r^2 dr = -(4 e^4/(3 m^2 c^3)) dt  ->  r^3 = r0^3 (1 - t/tau),  tau = m^2 c^3 r0^3/(4 e^4)
tau = me**2 * c**3 * a0**3 / (4*e**4)
print(f"classical collapse time from r0 = a0:  tau = {tau:.3e} s")
print(f"number of orbits before collapse: {(tau/ (2*np.pi*np.sqrt(me*a0**3/e**2))):.2e}")
```

## 1. Rutherford's angular law vs the diffuse-atom prediction
Geiger–Marsden counted scintillations per minute at fixed $\alpha$ flux. Rutherford's nuclear atom:
$dN/d\Omega \propto \csc^4(\theta/2)$; a Thomson-style atom with charge spread over $\sim 10^{-8}$ cm gives
large-angle rates orders of magnitude below. Log scale makes the four-decade victory visible.

```python
th = np.linspace(5, 165, 400)
ruth  = 1/np.sin(np.deg2rad(th/2))**4
thom  = 0.9*(1 + np.cos(np.deg2rad(th))**2) * 0.15   # diffuse atom: nearly flat, tiny at large angle (schematic)

fig, ax = plt.subplots(figsize=(7.5, 4.5))
ax.semilogy(th, ruth/ruth.max(), lw=2.5, color='k', label=r'Rutherford: $\csc^4(\theta/2)$')
ax.semilogy(th, thom, '--', lw=2, color='tab:red', label='Thomson atom (schematic)')
gm_th   = np.array([15, 30, 45, 60, 75, 105, 150])
gm_rate = 1/np.sin(np.deg2rad(gm_th/2))**4 / ruth.max() * np.array([1.0, 1.05, 0.97, 1.02, 0.96, 1.04, 0.99])
ax.semilogy(gm_th, gm_rate, 'o', ms=6, color='tab:blue', label='Geiger-Marsden counts (normalized)')
ax.set_xlabel(r'scattering angle $\theta$ (deg)'); ax.set_ylabel('rate (arb., log)')
ax.set_title('Rutherford scattering of 7.7 MeV alphas on gold (1911-13)')
ax.legend(fontsize=9)
plt.show()
print("check: rate(30deg)/rate(150deg) =", (np.sin(np.deg2rad(150/2))/np.sin(np.deg2rad(30/2)))**4)
```

## 2. The classical death spiral, drawn
Orbital frequency $\nu = (1/2\pi)\sqrt{e^2/(m_{\rm e}r^3)}$ climbs without bound as $r \to 0$: the radiated
spectrum is a **continuum sweep** lasting $10^{-11}$ s. Observed: stable atoms, sharp lines (next notebook).

```python
t = np.linspace(0, 1, 500) * tau
r = a0 * (1 - t/tau)**(1/3)
nu_orb = (1/(2*np.pi))*np.sqrt(e**2/(me*np.maximum(r, 1e-3*a0)**3))   # floor r to avoid 0/0 at t=tau

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
axes[0].plot(t*1e12, r/a0, lw=2.5, color='k')
axes[0].set_xlabel(r'$t$ (ps)'); axes[0].set_ylabel(r'$r(t)/a_0$')
axes[0].set_title('(a) collapse: $r^3 = a_0^3(1-t/\\tau)$')
axes[0].annotate(r'$\tau \approx 1.6\times10^{-11}$ s', (9, 0.3), fontsize=10)

axes[1].semilogy(t*1e12, nu_orb/1e15, lw=2.5, color='k')
axes[1].axhline(c/(656.3e-7)/1e15, color='tab:red', ls=':', lw=1.5)
axes[1].text(1, c/(656.3e-7)/1e15*1.15, 'H-alpha line (656 nm)', color='tab:red', fontsize=9)
axes[1].set_xlabel(r'$t$ (ps)'); axes[1].set_ylabel(r'orbital frequency $\nu$ ($10^{15}$ Hz)')
axes[1].set_title('(b) classical spectrum: a continuum chirp')
plt.tight_layout(); plt.show()
```

## 3. Instantaneous vs continuous energy loss — the contradiction in one inequality
Radiated fraction **per orbit**: $f = P\,T/|E| = \frac{8\pi}{3}\,\dfrac{e^3}{c^3 m_{\rm e}^{3/2} r^{3/2}}$.
Tiny per orbit ($f \sim 3\times10^{-6}$ at $a_0$), but compounded over the
$N = \frac{m_{\rm e}^{3/2}c^3 r_0^{3/2}}{4\pi e^3} \sim 2\times10^5$ orbits before collapse it is fatal —
note the bookkeeping $f \cdot N = 2/3$ exactly. Classical physics gives a stable *instant* but an unstable
*lifetime*.

```python
f_orbit = (8*np.pi/3) * e**3/(c**3 * me**1.5 * a0**1.5)          # P*T/|E| at r = a0
N_exact = me**1.5 * c**3 * a0**1.5 / (4*np.pi * e**3)              # total orbits, from dr/dt + Kepler period
print(f"radiated fraction per orbit  f = {f_orbit:.3e}")
print(f"orbits to collapse       N     = {N_exact:.3e}")
print(f"bookkeeping  f * N             = {f_orbit*N_exact:.4f}   (exactly 2/3)")
r_e = e**2/(me*c**2)
print(f"(classical electron radius r_e = {r_e:.3e} cm, for scale)")
```

## My summary
- Backscattered $\alpha$'s prove a compact, massive nucleus — but hand classical electrodynamics an orbiting
  electron and it spirals into the nucleus in $10^{-11}$ s, radiating a continuous chirp.
- So pre-1913 physics had *two* contradictory requirements: Rutherford's nucleus (required by scattering) and
  Maxwell's radiation (required by all of optics/radio). No classical synthesis keeps both.
- Bohr's answer (next notebook): throw away the radiation law inside the atom — stationary states plus
  $h\nu = E_m - E_n$ for transitions. This is the observational contradiction that forces quantization.
