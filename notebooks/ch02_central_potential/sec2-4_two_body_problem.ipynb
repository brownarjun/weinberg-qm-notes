# Section 2.4 — The Two-Body Problem (Weinberg §2.4)

**What this section claims**
- For H = p₁²/2m₁ + p₂²/2m₂ + V(|x₁−x₂|), CM coordinates X = (m₁x₁+m₂x₂)/M and relative x = x₁−x₂ (M = m₁+m₂) decouple the kinetic energy: H = P²/2M + p²/2μ + V(r) with reduced mass μ = m₁m₂/(m₁+m₂).
- The bound-state spectrum of a two-body system is that of one particle of mass μ in V(r); CM motion is a free particle.
- Hydrogen vs deuterium: identical formula with μ ⇒ all lines shifted by μ_D/μ_H ≈ 1 + 2.7×10⁻⁴ (≈ 1.8 Å on Hα) — Urey's 1932 discovery of deuterium.
- Finite-nuclear-mass corrections are O(m_e/m_p); μ → m_e as the nucleus gets heavy.


## Kinetic-energy separation in CM/relative coordinates (symbolic)

Chain rule on a test function g(X(x₁,x₂), x(x₁,x₂)) shows the mixed second derivatives cancel exactly:


```python
import sympy as sp

m1, m2 = sp.symbols('m1 m2', positive=True)
M, mu = m1+m2, m1*m2/(m1+m2)
# chain rule (stated in the text):  d/dx1 = (m1/M) d/dX + d/dx,  d/dx2 = (m2/M) d/dX - d/dx
# expand both sides in the independent partials g_XX, g_Xx, g_xx and compare coefficients
gXX, gXx, gxx = sp.symbols('g_XX g_Xx g_xx')
lhs = sp.expand((1/(2*m1))*((m1/M)**2*gXX + 2*(m1/M)*gXx + gxx)
            + (1/(2*m2))*((m2/M)**2*gXX - 2*(m2/M)*gXx + gxx))
rhs = gXX/(2*M) + gxx/(2*mu)
print("LHS =", sp.simplify(lhs))
print("LHS - RHS =", sp.simplify(lhs - rhs), "  =>  p1^2/2m1 + p2^2/2m2 = P^2/2M + p^2/2mu")
```

## Reduced-mass expansion for m₁ ≪ m₂ (symbolic)


```python
eps_ = sp.symbols('eps_', positive=True)
mu_exp = sp.series(m1*m2/(m1+m2), m2, sp.oo, 3)
print("mu = m1 m2/(m1+m2)  expanded at m2 -> oo:", mu_exp)
print()
print("=> mu = m1 (1 - m1/m2 + ...)  :  nuclear-mass corrections are O(m_e/m_p) ~ 1/1836")
```

## Numerics: H vs D spectral lines (Weinberg's isotope-shift example)


```python
import numpy as np, sys
sys.path.insert(0, '..')
from weinberg_constants import me, mp, md, Ryd, h, c

mu_H = me*mp/(me+mp); mu_D = me*md/(me+md)
print(f"mu_H = {mu_H/me:.6f} m_e,  mu_D = {mu_D/me:.6f} m_e,  ratio = {mu_D/mu_H:.6f} (+{(mu_D/mu_H-1)*1e4:.2f} x10^-4)")

def wavelength(n, mu_):
    R_M = Ryd*mu_/me/(h*c)             # cm^-1  (Ryd*mu/me is the finite-mass Rydberg energy)
    return 1.0/(R_M*(1/4 - 1/n**2))    # n -> 2, cm

print(f"{'line':>8} {'lambda_H (A)':>14} {'lambda_D (A)':>14} {'shift (A)':>10}")
for n, name in [(3,'H-alpha'), (4,'H-beta '), (5,'H-gamma')]:
    lam_H, lam_D = wavelength(n, mu_H)*1e8, wavelength(n, mu_D)*1e8
    print(f"{name:>8} {lam_H:14.3f} {lam_D:14.3f} {lam_D-lam_H:10.3f}")
print("Urey (1932): deuterium discovered via this ~1.8 A Balmer-line doublet")
```

## Numerics: how μ approaches m_e; the muonic-hydrogen remark


```python
import matplotlib.pyplot as plt
Mnuc = np.geomspace(0.5, 250, 300)*mp       # nuclear mass from 0.5 mp to 250 mp
mu_of = me*Mnuc/(me+Mnuc)
fig, ax = plt.subplots(figsize=(7, 4))
ax.semilogx(Mnuc/me, mu_of/me, lw=2)
ax.axhline(1, color='k', ls=':', lw=1); ax.set_xlabel('M_nucleus / m_e'); ax.set_ylabel('mu / m_e')
ax.set_title('reduced mass:  mu/m_e = M/(M+m_e)  ->  the spectrum is set by mu, not m_e')
# a muon (m ~ 207 m_e) as 'nucleus': Bohr radius shrinks by ~186 -> proton-radius physics
mu_muonic = me*(207*me)/(me+207*me)
ax.annotate(f'muonic H: mu = {mu_muonic/me:.1f} m_e\n(a0 -> {me/mu_muonic:.0f}x smaller)',
            xy=(207, mu_muonic/me), xytext=(12, 0.75), arrowprops=dict(arrowstyle='->'), fontsize=9)
plt.tight_layout(); plt.show()
```
